# Building response to earthquake

### Solving ordinary differential equation of dynamic motion involving mass, damping and spring in response to earthquake

![](https://github.com/TUDelft-MUDE/source-files/raw/main/file/1Ddynamic.png)


A building, considered here as a rigid mass, is subject to an earthquake. The foundation of the building, interacting with the ground, is represented by a spring and a dashpot (damper).

The dynamic motion of the building is defined by the following differential equation:

$$ k \left(u_b(t)-u_g(t) \right) + c \left(\frac{d u_b}{d t} - v_g(t) \right) + m \left(\frac{d^2 u_b}{d t^2} \right) = F $$

where:
*   $u_b$ = horizontal displacement of the building
*   $u_g$ = horizontal displacement of the ground (= earthquake outcrop motion)
*   $v_g$ = $du_g / dt$ = horizontal velocity of the ground
*   $t$ = time
*   $k$ = (shear) stiffness between the building foundation and the ground
*   $c$ = damping between the building foundation and the ground
*   $m$ = mass of the building
*   $F$ = external force acting on the building

In the remainder of this assignment we consider $F$ = 0

```python
from itertools import islice


def readfile(filename):
    """Read data from file and return the data as a 1-D numpy array."""
    values = []
    with open(filename) as f:
        for line in islice(f, 3, None):  # Skip first 3 rows
            values.append(np.fromstring(line, sep=" "))
    return np.concatenate(values)


ag = readfile("Earthquake_acceleration.txt")
vg = readfile("Earthquake_velocity.txt")
ug = readfile("Earthquake_displacement.txt")

num = vg.shape[0]  # number of datapoints in the earthquake record
dt = 0.005  # time interval between subsequent data points
t = np.linspace(start=0, stop=dt * num, num=num)
```

```python
# Forward Difference scheme to differentiate ground acceleration from given velocity
### YOUR CODE HERE ###
def forward_difference(v, dt):
    """Compute the forward difference"""
    a_g = np.zeros(num - 1)
    for i in range(num - 1):
        a_g[i] = (v[i + 1] - v[i]) / dt
    return a_g

ag_calculated = forward_difference(vg, dt)
### YOUR CODE HERE ###
```

```python
# Trapezoidal Rule to integrate ground displacement from given velocity

### YOUR CODE HERE ###

def trapezoidal_rule(v, dt):
    """Compute the trapezoidal rule"""
    u_g = np.zeros(num)
    for i in range(1, num):
        u_g[i] = u_g[i - 1] + (v[i] + v[i - 1]) * dt / 2
    return u_g

ug_calculated = trapezoidal_rule(vg, dt)

### YOUR CODE HERE ###
### YOUR CODE HERE ###
```

## Part 2: Numerical solution of the dynamic problem

Now, we proceed with the numerical solution of the above differential equation for the dynamic motion of the building $u_b$. Since the second derivative already requires two subsequent timesteps, we use the Central Difference scheme both for the first and second derivative to obtain maximum accuracy. Note that the ground acceleration is not used in the differential equation and the numerical solution; only the ground displacement and velocity! (Why?)

### Task 2.1

The original differential equation is:

$$
k(u_b-u_g)
+
c\left(\frac{du_b}{dt}-v_g\right)
+
m\frac{d^2u_b}{dt^2}
=0
$$

Using the Central Difference scheme for the first derivative:

$$
\frac{du_b}{dt}
\approx
\frac{u_b^{i+1}-u_b^{i-1}}{2\Delta t}
$$

and for the second derivative:

$$
\frac{d^2u_b}{dt^2}
\approx
\frac{u_b^{i+1}-2u_b^i+u_b^{i-1}}{\Delta t^2}
$$

Substituting these into the original differential equation gives:

$$
k(u_b^i-u_g^i)
+
c\left(
\frac{u_b^{i+1}-u_b^{i-1}}{2\Delta t}
-v_g^i
\right)
+
m\left(
\frac{u_b^{i+1}-2u_b^i+u_b^{i-1}}{\Delta t^2}
\right)
=0
$$

Collecting the terms with $u_b^{i+1}$:

$$
\left(
\frac{m}{\Delta t^2}
+
\frac{c}{2\Delta t}
\right)u_b^{i+1}
=
\left(
\frac{2m}{\Delta t^2}-k
\right)u_b^i
+
\left(
\frac{c}{2\Delta t}
-
\frac{m}{\Delta t^2}
\right)u_b^{i-1}
+
ku_g^i
+
cv_g^i
$$

Therefore, the building displacement at the next time step is:

$$
\boxed{
u_b^{i+1}
=
\frac{
\left(\frac{2m}{\Delta t^2}-k\right)u_b^i
+
\left(\frac{c}{2\Delta t}-\frac{m}{\Delta t^2}\right)u_b^{i-1}
+
ku_g^i
+
cv_g^i
}{
\frac{m}{\Delta t^2}+\frac{c}{2\Delta t}
}
}
$$

### Task 2.2

Because this is an initial value problem, we use one zeroth-order initial condition and one first-order initial condition at the start of the motion.

If it were a boundary value problem, we could instead use two independent zeroth-order conditions at different points.

For this problem, the initial displacement is:

$$
u_b(0) = 0
$$

and the initial velocity is:

$$
\frac{du_b}{dt}(0) = v_b(0) = 0
$$

So the two required initial conditions are:

$$
u_b(0)=0
$$

and

$$
v_b(0)=0
$$

For the numerical scheme:

$$
u_b^0 = 0
$$

and, because the initial velocity is zero,

$$
\frac{u_b^1-u_b^0}{\Delta t}=0
$$

therefore:

$$
u_b^1=u_b^0=0
$$

So the numerical solution starts with zero displacement and zero velocity.

### Implementation of the numerical solution

```python
def solution(k, c, m, num):
    """Solve the ODE for the building displacement in time
    
    Parameters
    ----------
    k : float
        Stiffness of the building
    c : float
        Damping of the building
    m : float
        Mass of the building
    num : int
        Number of time steps to solve for
    """
    u = np.zeros([num])  # Initial conditions at t=0 automatically satisfied

    ### YOUR CODE HERE ###
    for i in range(1, num-1):
        u[i + 1] = (
            (2 * m / dt**2 - k) * u[i]
            + (c / (2 * dt) - m / dt**2) * u[i - 1]
            + k * ug[i]
            + c * vg[i]
        ) / (
            m / dt**2 + c / (2 * dt)
        )
    ### YOUR CODE HERE ###
    return u


ub = solution(k, c, m, num)
```

### Task 2.5

The influence of the parameters can be understood from the dynamic equation

$$
k(u_b-u_g)
+
c(v_b-v_g)
+
m a_b
=0
$$

#### Stiffness $k$

- Increasing $k$ by a factor 10 makes the building much stiffer.
- The building follows the ground displacement more closely because a large difference between $u_b$ and $u_g$ produces a large restoring force.
- The natural frequency increases:

$$
\omega_n = \sqrt{\frac{k}{m}}
$$

- Therefore, oscillations become faster.

When $k$ is decreased by a factor 10, the building is more flexible. It reacts more slowly to the movement of the ground and the difference between building and ground displacement can become larger.

#### Damping $c$

- Increasing $c$ by a factor 10 strongly reduces oscillations.
- The building loses vibration energy more quickly and its motion becomes smoother.
- The building velocity is forced to follow the ground velocity more closely.

When $c$ is decreased by a factor 10, there is much less damping. Oscillations therefore continue for a longer time and can have a larger amplitude.

#### Mass $m$

- Increasing $m$ by a factor 10 increases the inertia of the building.
- The building therefore reacts more slowly to the ground motion and tends to lag behind it.
- The natural frequency decreases because:

$$
\omega_n = \sqrt{\frac{k}{m}}
$$

- A larger mass can therefore result in slower oscillations.

When $m$ is decreased by a factor 10, the building has less inertia and can react more quickly to the ground motion. The natural frequency increases.

#### Oscillations

Oscillations occur when the restoring force from the stiffness causes the building to move back towards its equilibrium position, but because of its mass it overshoots this position.

The amount of damping determines how quickly these oscillations disappear.

Therefore:

- high $k$ → faster oscillations
- low $k$ → slower and more flexible response
- high $c$ → fewer oscillations
- low $c$ → stronger and longer-lasting oscillations
- high $m$ → slower response and lower oscillation frequency
- low $m$ → faster response and higher oscillation frequency
